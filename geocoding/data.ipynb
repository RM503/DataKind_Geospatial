# **Clean up location data**

We make the distributor location data uniform and enforce schema for consitent use later.

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("distributor_locations_priority_geocoded.csv", index_col=[0])

df
```

## **Rename columns**

```python
col_names = df.columns.tolist()
new_col_names = (
    df.columns
      .str.lower()
      .str.strip()
      .str.replace(" ", "_")
)

col_name_mapping = {
    k: v for k, v in zip(col_names, new_col_names)
    if k != "#"
}

col_name_mapping["#"] = "distributor_id"

df = df.rename(columns=col_name_mapping)
df["distributor_id"] = np.arange(1, len(df) + 1)
```

```python
df["locations_cleaned"]
```

```python
df["addresses"] = (
    df["locations_cleaned"]
    .str.replace(",", "", regex=False)
    .str.replace(r"\s+", " ", regex=True) # more than one whitespace
    .str.strip()
)

df = df.drop(["locations", "locations_cleaned"], axis=1)
```
